<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс BankAccount в C#, который будет представлять
информацию об учетных записях в банке. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

Требования к базовому классу BankAccount:
    • Атрибуты: Номер счета (AccountNumber), Баланс (Balance), Тип счета
(AccountType).
    • Методы:
        o GetInfo(): метод для получения информации о счете в виде строки.
        o Deposit(decimal amount): метод для внесения денег на счет.
        o Withdraw(decimal amount): метод для снятия денег со счета.
Требования к производным классам:
1. Сберегательный счет (SavingsAccount): Должен содержать дополнительные
атрибуты, такие как Процентная ставка (InterestRate). Метод Deposit() должен
быть переопределен для добавления процентов к сумме вклада при
внесении денег на счет.
2. Текущий счет (CheckingAccount): Должен содержать дополнительные
атрибуты, такие как Лимит овердрафта (OverdraftLimit).
Метод Withdraw() должен быть переопределен для проверки и применения
лимита овердрафта при снятии денег со счета.
3. Инвестиционный счет (InvestmentAccount) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Список активов
(AssetsList). Метод GetInfo() должен быть переопределен для включения
информации о списках активов в описании счета.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
public class BankAccount {
    private static int _totalAccounts;

    private string _accountNumber;
    private  string _accountType;
    private decimal _balance;    
    private string _ownerName;
    private string _conatactNumber;

    protected string GetAccountNumber() { return _accountNumber; }
    protected void SetAccountNumber(string newNumber) { _accountNumber = newNumber; }

    protected string GetAccountType() { return _accountType; }
    protected void SetAccountType(string newType) { _accountType = newType; }

    protected decimal GetBallance() { return _balance; }
    protected void SetBallance(decimal newBallance) { _balance = newBallance; }

    protected string GetOwnerName() { return _ownerName; }
    protected void SetOwnerName(string name) { _ownerName = name; }

    protected string GetContactNumber() { return _conatactNumber; }
    protected void SetContactNumber(string number) { _conatactNumber = number; }

    public BankAccount(string AccNum, string AccTp, decimal val) {
        SetAccountNumber(AccNum);
        SetAccountType(AccTp);
        SetBallance(val);
        _totalAccounts++;
    }

    public virtual void GetInfo() {
        Console.WriteLine($"Account number: {GetAccountNumber()},\nAccount type: {GetAccountType()},\nBalance: {GetBallance()}.");
    }

    public virtual void Deposit(decimal amount) {
        SetBallance(GetBallance() + amount);
        Console.WriteLine($"{amount} has been deposited to your account {GetAccountNumber()}. New balance: {GetBallance()}.");
    } 

    public virtual void Withdraw(decimal amount) {
        SetBallance(GetBallance() - amount);
        Console.WriteLine($"{amount} has been withdraw from your account {GetAccountNumber()}. New balance: {GetBallance()}.");
    }

    public virtual void DisplayTotalAccounts() {
        Console.WriteLine($"Total accounts in bank: {_totalAccounts}.");
    }
}

public class SavingsAccount : BankAccount {
    private static int _totalSavings;
    
    private decimal _interestRate;

    protected decimal GetInterestRate() { return _interestRate; }
    protected void SetInterestRate(decimal newInterest) { _interestRate = newInterest; }


    public SavingsAccount(string AccNum, decimal val, decimal rate) : base(AccNum, "Savings", val) {
        SetInterestRate(rate);
        _totalSavings++;
    }

    public override void Deposit(decimal amount) {
        decimal interest = amount * GetInterestRate();
        base.Deposit(interest + amount);
        Console.WriteLine($"Interest accured: {interest}.");
    }

    public override void GetInfo() {
        base.GetInfo();
        Console.WriteLine($"Interest rate: {GetInterestRate()}.");
    }

    public override void DisplayTotalAccounts() {
        Console.WriteLine($"Total Savings Accounts in bank: {_totalSavings}.");
    }
}

public class CheckingAccount : BankAccount {
    private static int _totalChecking;

    private decimal _overdraftLimit;

    protected decimal GetOverdraftLimit() { return _overdraftLimit; }
    protected void SetOverdraftLimit(decimal newLimit) { _overdraftLimit = newLimit; }

    public CheckingAccount(string AccNum, decimal val, decimal lim) : base(AccNum, "Checking", val) {
        SetOverdraftLimit(lim);
        _totalChecking++;
    }

    public override void Withdraw(decimal amount) {
        if (GetBallance() + GetOverdraftLimit() >= amount) {
            base.Withdraw(amount);

            if (GetBallance() < 0) Console.WriteLine($"Warning! Overdraft limit is over.\nAvailable limit: {GetBallance() + GetOverdraftLimit()}.");
        } else {
            Console.WriteLine("Warning! Overdraft limit is over.");
        }
    }

    public override void GetInfo() {
        base.GetType();
        Console.WriteLine($"Overdraft limit: {GetOverdraftLimit()}.");
    }

    public override void DisplayTotalAccounts() {
        Console.WriteLine($"Total Cheching Accounts in bank: {_totalChecking}.");
    }
}

public class InvestmentAccount : BankAccount {
    private static int _totalInvestment;
    
    private List<string> _assetsList;

    protected List<string> GetAssetsList() { return _assetsList; }
    protected void SetAssetsList(List<string> newList) { _assetsList = newList; }

    public InvestmentAccount(string AccNum, decimal val, List<string> asL) : base(AccNum, "Investment", val) {
        SetAssetsList(asL);
        _totalInvestment++;
    }

    public override void GetInfo() {
        string list = string.Join(", ", GetAssetsList());
        base.GetInfo();
        Console.WriteLine("Assets list: " + list);
    }

    public void AddAsset(string asset) {
        GetAssetsList().Add(asset);
        Console.WriteLine($"Add asset: {asset} on account {GetAccountNumber()}.");
    }

    public void RemoveAsset(string asset)
    {
        if (GetAssetsList().Contains(asset))
        {
            GetAssetsList().Remove(asset);
            Console.WriteLine($"Remove asset: {asset} from account {GetAccountNumber()}.");
        }
        else
        {
            Console.WriteLine($"Asset {asset} not found on account {GetAccountNumber()}.");
        }
    }

    public override void DisplayTotalAccounts() {
        Console.WriteLine($"Total Investment Account in bank: {_totalInvestment}.");
    }

}

public class AccountCollection<T> where T : BankAccount {
    private List<T> _acconts = new List<T>();

    public void Add(T acc) { _acconts.Add(acc); }
    public void Remove(T acc) { _acconts.Remove(acc); }

    public void DisplayAccounts() {
        foreach (T acc in _acconts) {
            acc.GetInfo();
            Console.WriteLine();
        }
    }
}

BankAccount account1 = new BankAccount("1234567890", "Main", 1000);
InvestmentAccount investmentAccount = new InvestmentAccount("1122334455", 1500, new List<string> { "Apple stocks", "Obligation" });

account1.GetInfo();
account1.Deposit(500);
account1.Withdraw(200);

Console.WriteLine();

investmentAccount.GetInfo();
investmentAccount.AddAsset("Tesla stocks");
investmentAccount.RemoveAsset("Obligation");
investmentAccount.GetInfo();

Console.WriteLine();
Console.WriteLine("-------------------------------------------------------");
Console.WriteLine();

AccountCollection<BankAccount> accs = new AccountCollection<BankAccount>();
accs.Add(account1);
accs.Add(new SavingsAccount("88005553535", 40000, 15));
accs.DisplayAccounts()
```
